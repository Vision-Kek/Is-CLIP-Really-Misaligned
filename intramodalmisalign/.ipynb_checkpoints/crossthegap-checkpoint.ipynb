```python
def sample_random_idcs(nel, ratio=1/25000):
    return torch.randint(high=nel, size=(int(nel*ratio),))
```

```python
random_idcs = sample_random_idcs(simmat_dino_l.numel())
```

```python
px.histogram({'dino-large': simmat_dino_l.view(-1)[random_idcs].cpu(),
              'clip-base': simmat_clip_b.view(-1)[random_idcs].cpu().float()},
            barmode='overlay')
```

```python
px.scatter(x=simmat_dino_l.view(-1)[random_idcs].cpu(), 
           y=simmat_clip_b.view(-1)[random_idcs].cpu(), title='dino-l vs clib-b')
```
