```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import torch
import torch.nn.functional as F
from PIL import Image
import os
import importlib
import json
from tqdm import tqdm
from torch.utils.data import Dataset, DataLoader
import clip
import plotly.express as px

from cddataset import CatDogDataset
from algs import calc_simmat, fs_classification, measure_retrieval_map, measure_text_to_image_retrieval_map, measure_zero_shot_classification_acc
import feat_extraction

device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
dataset = CatDogDataset()
print(len(dataset))
dataloader = DataLoader(dataset, shuffle=False, batch_size=128, collate_fn=lambda x: x)
```

```python
clip_model, clippreprocess = clip.load("ViT-B/16", device=device)
with torch.no_grad():
    text_input = clip.tokenize(['a photo of a cat', 'a photo of a dog']).to(device)
    temb = clip_model.encode_text(text_input)
    temb = temb.to(device)

@torch.no_grad
def extract_features(model_id):
    if model_id=='clip_b':
        embs = feat_extraction.loop_clip(clip_model, clippreprocess, dataloader)
    elif model_id=='dino_b':
        dino_b = feat_extraction.DinoHF(device)
        embs = dino_b.loop(dataloader)
    elif model_id=='dino_l':
        embs = feat_extraction.loop_dinov2_l(dataloader)
    elif model_id=='dinov3_l':
        embs = feat_extraction.loop_dinov3_l(dataloader)
    else: raise ValueError(f'no such {model_id=}')
    return F.normalize(embs, dim=-1)

def save_embeds(embs, pt_name):
    save_dir = os.path.join(dataset.catdog_dataset, 'embeds')
    torch.save(embs, os.path.join(save_dir, pt_name))

def load_embeds(pt_name):
    load_dir = os.path.join(dataset.catdog_dataset, 'embeds')
    return torch.load(os.path.join(load_dir, pt_name))
```

```python
clip_b_embs = load_embeds('clip_b_16.pt')
dino_b_embs = load_embeds('dino_b.pt')
dinov3_l_embs = load_embeds('dinov3_l.pt')
```

```python
# # Measure text query classification accuracy
def measure_zero_shot_classification_acc(img_embs, text_embs, labels):
    pred = (img_embs.to(device) @ text_embs.to(device).T).argmax(dim=-1)
    acc = (pred == labels).float().mean()
    return acc
```

```python
labels = torch.tensor(dataset.labels,device=device)
measure_zero_shot_classification_acc(clip_b_embs, temb, labels)
```

### For illustration purposes, for a fixed img-txt sim, select the image pair with the lowest sim and highest sim
This should demonstrate the point that *there is a valid reason* for different distances, because the embedding doesn't (and shouldn't) capture the target concept only

```python
def show_different_img_sims_for_same_txt_sim(clip_b_embs, temb):
    temb = F.normalize(temb, dim=-1).cpu()
    img_txt_sim = clip_b_embs @ temb.T # N, 2
    img_img_sim = clip_b_embs @ clip_b_embs.T # N, N
    # 1. choose a fixed small img-txt sim range (bin)
    cat_sim = img_txt_sim[:,0].float()
    hist = torch.histogram(cat_sim) # histogram of sims to cat
    bins = hist.bin_edges[:100][hist.hist >= 2] # filter bins with less than 2 samples
    selected_bin = len(bins) // 2 + 14 # adjust here to select a bin
    select_criterion = (bins[selected_bin+1]>cat_sim) & (bins[selected_bin] < cat_sim) # within bin
    clip_embs_equal_img_txt = clip_b_embs[select_criterion] # select embeds within bin
    # 2. calculate the img_img_sims among these selected samples with equal img_txt_sim
    _img_txt_sim = clip_embs_equal_img_txt @ temb.T # N, 2
    _img_img_sim = clip_embs_equal_img_txt @ clip_embs_equal_img_txt.T # N, N
    # 3. find min and max img_img_sim
    _img_img_sim += -torch.eye(len(_img_img_sim))+torch.eye(len(_img_img_sim))*_img_img_sim[~torch.eye(len(_img_img_sim),dtype=bool)].mean()#_img_img_sim[~torch.eye(len(_img_img_sim),dtype=bool)] # exclude diagonal
    _min, _max = _img_img_sim.flatten().topk(k=1,largest=False).values[0], torch.max(_img_img_sim)
    # 4. convert to image indices in dataset: two pairs: (min1,min2), (max1,max2)
    min_idx_in_sel = (_img_img_sim==_min).nonzero()[0]
    max_idx_in_sel = (_img_img_sim==_max).nonzero()[0]
    min_idx = select_criterion.nonzero()[min_idx_in_sel].squeeze()
    max_idx = select_criterion.nonzero()[max_idx_in_sel].squeeze()
    # verify
    print(f"txt min={cat_sim[min_idx]},  max={cat_sim[max_idx]}")
    print(f"img {_min=},{_max=}")
    assert img_img_sim[min_idx[0],min_idx[1]] == _min and img_img_sim[max_idx[0],max_idx[1]] == _max
    # 5. visualize
    print('min')
    display(dataset[min_idx[0]][0])
    display(dataset[min_idx[1]][0])
    print('max')
    display(dataset[max_idx[0]][0])
    display(dataset[max_idx[1]][0])

show_different_img_sims_for_same_txt_sim(clip_b_embs, temb)
```

```python
def simmat_mse(simmat1, simmat2):
    d = (simmat1 - simmat2).square().mean()
    d_n = ((simmat1-simmat1.mean()) - (simmat2-simmat2.mean())).square().mean()
    return d, d_n
    
print(simmat_mse(torch.randn(10000), torch.randn(10000)))
print(simmat_mse(simmat_dinob/simmat_dinob.std(), simmat_dino_l/simmat_dino_l.std()))
print(simmat_mse(simmat_clip_b/simmat_clip_b.std(), simmat_dino_l/simmat_dino_b.std()))
print(simmat_mse(simmat_clip_b/simmat_clip_b.std(), simmat_dino_l/simmat_dino_l.std()))
```
