```python
def fwd_texts(classnames, tokenizer):
    with torch.no_grad():
        zeroshot_weights = []
        for classname in classnames:
            texts = f"a photo of a {classname}" #format with class
            texts = tokenizer.tokenize(texts).cuda() #tokenize
            class_embeddings = model.encode_text(texts)[:,:1024] #embed with text encoder
            class_embeddings /= class_embeddings.norm(dim=-1, keepdim=True)
            class_embedding = class_embeddings.mean(dim=0)
            class_embedding /= class_embedding.norm()
            zeroshot_weights.append(class_embedding)
        zeroshot_weights = torch.stack(zeroshot_weights, dim=0)
    return zeroshot_weights
    
text_embeds = fwd_texts(imagenet_names, tokenizer)
text_embeds.shape
```

```python
from tqdm import tqdm
@torch.no_grad()
def fwd_images(dataloader):
    image_embeds = []
    for batch in tqdm(dataloader):
        image_tensor = batch[0].to(device).expand(-1,3,-1,-1)
        # https://github.com/facebookresearch/dinov2/blob/main/dinov2/hub/text/vision_tower.py#L151
        backbone_class_tokens, backbone_patch_tokens, register_tokens = model.visual_model.get_backbone_features(image_tensor)
        image_tokens = model.visual_model.head(torch.cat([backbone_class_tokens.unsqueeze(1), 
                                                          register_tokens, 
                                                          backbone_patch_tokens],
                                                         dim=1))
        image_class_tokens, image_patch_tokens = image_tokens[:, 0], image_tokens[:, model.visual_model.num_register_tokens + 1:]

        image_embeds.append(image_class_tokens.cpu())
    return torch.cat(image_embeds)
```
