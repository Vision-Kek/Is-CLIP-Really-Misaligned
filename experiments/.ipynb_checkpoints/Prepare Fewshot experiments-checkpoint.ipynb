**Now enable reuse of embeddings from CTG**:
- In GDA, we want a single n,d tensor per dataset with all embeddings in the .pt
- and n corresponding names in the pkl

Since we don't need the val features unless we do hyperparm grid search,
1. Keep image paths in `imagenames.pkl`
2. Merge train-> <-test *embeds* from CTG into 1 tensor
3. Insert zeros for *embeds* that are missing in the center (correspond to val)

```python
def merge(dataset, model):
    # 1. Keep image paths in `imagenames.pkl`
    gda_n = pickle.load(open(f'/home/jonas/workspace/ICLR24/data/image_features/{dataset}/image_names.pkl','rb'))
    n_total = len(gda_n)
    # 2. Merge train-> <-test *embeds* from CTG into 1 tensor
    ctg_train_x = torch.load(f'/home/jonas/workspace/Cross-the-Gap/data/image_features/{dataset.replace("-","")}/{model}/train/image_features.pt')
    ctg_test_x = torch.load(f'/home/jonas/workspace/Cross-the-Gap/data/image_features/{dataset.replace("-","")}/{model}/test/image_features.pt')
    n_train = len(ctg_train_x)
    n_test = len(ctg_test_x)
    n_val = n_total - n_train - n_test
    print(n_train, n_val, n_test)
    # 3. Insert zeros for *embeds* that are missing in the center (correspond to val)
    merged = torch.cat([ctg_train_x, torch.zeros(n_val, ctg_train_x.shape[1]), ctg_test_x])
    return merged

def check_equal_gda_feats_ctg_feats(merged):
    cossim = torch.nn.CosineSimilarity(dim=-1)
    # extracted by GDA
    model = 'ViT-B_16'
    gda_x = torch.load(f'/home/jonas/workspace/ICLR24/data/image_features/{dataset}/{model}.pt').cpu()
    assert len(gda_x)==len(merged), (len(gda_x),len(merged))
    sim = cossim(gda_x, merged)
    print(sim)
    import plotly.express as px
    display(px.histogram(sim,nbins=300))
 
merged = merge(dataset, 'ViT-B16')
print(merged.shape)
check_equal_gda_feats_ctg_feats(merged)
```

**Now create for each model the merged tensor and write it to the .pt in ICLR24/data**

```python
import os

ctg_model_folders = os.listdir(f'/home/jonas/workspace/Cross-the-Gap/data/image_features/{dataset}')
for model in ctg_model_folders:
    merged = merge(dataset, model)
    print(model, merged.shape)
    # write
    gda_out_file = f'/home/jonas/workspace/ICLR24/data/image_features/{dataset}/{model}.pt'
    torch.save(merged, gda_out_file)
    #names.append(pickle.load(open(f'/home/jonas/workspace/ICLR24/data/image_features/{dataset}/image_names.pkl','rb')))
```
